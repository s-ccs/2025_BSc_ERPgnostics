# ResNet18 CV with inverse-sort and polarity augmentation

This Week-23 experiment uses the same Label Studio annotations and source datasets as `notebooks/week_21/resnet18_labeled_erp_cv.ipynb`, but materializes fixed 200-trial ERP images. Each mod-split image is rendered as four variants: reference, inverse sort, inverse polarity, and inverse sort plus inverse polarity. The variant operation is applied before `zscore -> Gaussian smoothing -> resize`.

```julia
import Pkg

function find_repo_root(start::AbstractString = pwd())
    candidate = start
    for _ in 1:8
        if isdir(joinpath(candidate, "notebooks")) && isdir(joinpath(candidate, "scripts"))
            return candidate
        end
        candidate = dirname(candidate)
    end
    error("Could not locate repo root from ", start)
end

REPO_ROOT = find_repo_root()
WEEK23 = joinpath(REPO_ROOT, "notebooks", "week_23")
OUTPUT_DIR = joinpath(WEEK23, "outputs", "augmentation_inverse_sort_polarity")

Pkg.activate(joinpath(REPO_ROOT, "notebooks", "model_test"))
using CSV, DataFrames, JSON3
REPO_ROOT
```

Optional overrides before running: `WEEK23_RESNET18_EPOCHS`, `WEEK23_RESNET18_FOLDS`, `WEEK23_TARGET_TRIALS`, and `WEEK23_NO_CLASS_CHUNKS_PER_ORIGIN`. The default target trial count is 200.

```julia
get!(ENV, "WEEK23_TARGET_TRIALS", "200")

run(Cmd(`julia --project=notebooks/model_test $(joinpath(WEEK23, "augmentation_inverse_sort_polarity.jl"))`; dir = REPO_ROOT))
```

```julia
metrics_summary = CSV.read(joinpath(OUTPUT_DIR, "metrics_summary.csv"), DataFrame)
fold_metrics = CSV.read(joinpath(OUTPUT_DIR, "fold_metrics.csv"), DataFrame)
fold_classes = CSV.read(joinpath(OUTPUT_DIR, "fold_distribution_pattern_class.csv"), DataFrame)
fold_augmentations = CSV.read(joinpath(OUTPUT_DIR, "fold_distribution_augmentation.csv"), DataFrame)
run_config = JSON3.read(read(joinpath(OUTPUT_DIR, "run_config.json"), String))

config_subset = (
    target_trials       = run_config.target_trials,
    target_size         = run_config.target_size,
    lowpass_sigma       = run_config.lowpass_sigma,
    lowpass_kernel_size = run_config.lowpass_kernel_size,
    n_labeled_rows_used = run_config.n_labeled_rows_used,
    n_base_mod_images   = run_config.n_base_mod_images,
    n_augmented_images  = run_config.n_augmented_images,
    k_folds             = run_config.k_folds,
    nepochs             = run_config.nepochs,
)

display(metrics_summary)
config_subset
```
