# Overview of Sort Variables for 8bit Game ERP Images

This notebook provides a visual overview of the available sort variables for the new 8bit game data.

Each sort variable gets its own figure. There is **no export** in this notebook.


## Processing

The ERP images are built with the same base pipeline as the other real datasets:
- post-stimulus segment only
- sorting by the selected sort variable
- `zscore_timepoints(...)` from the utils
- Gaussian low-pass with `gaussian_kernel(...)` from the utils
- **no resize**
- asymmetric zero-anchored colorbar via `clipped_color_stats_quantile_zero_ticks(...)`

The figures are organized with `trial_type` as rows and randomly selected channels as columns.


```julia
import Pkg

week15_dir = if isfile(joinpath(pwd(), "sort_variable_overview_8bit.ipynb"))
    pwd()
else
    joinpath(pwd(), "notebooks", "week_15")
end

Pkg.activate(joinpath(week15_dir, "..", "model_test"))

using CairoMakie
using DataFrames

include(joinpath(week15_dir, "sort_variable_overview_helpers.jl"))
using .Week15SortVariableOverview

println("OpenNeuro repo dir: ", OPENNEURO_REPO_DIR)
println("Derived 8bit dir: ", OPENNEURO_DERIVED_DIR)
```

```julia
data_bundle = load_8bit_overview_data()
sort_summary_df = summarize_sort_columns(data_bundle)

println("8bit EEG shape: ", data_bundle.full_shape)
println("Events rows: ", size(data_bundle.events, 1))
println("Game trial types: ", data_bundle.trial_types)

sort_summary_df
```

```julia
N_RANDOM_CHANNELS = 4
RANDOM_SEED = 15
PLOT_SORT_COLUMNS = collect(DEFAULT_OVERVIEW_SORT_COLUMNS)

println("Plot sort columns: ", PLOT_SORT_COLUMNS)
println("Random channels per figure: ", N_RANDOM_CHANNELS)
println("Base seed: ", RANDOM_SEED)
```

## onset_s

Sorting by event onset in seconds.


```julia
preview_onset = build_sortvar_preview(data_bundle;
    sort_col = :onset_s,
    n_channels = N_RANDOM_CHANNELS,
    seed = RANDOM_SEED,
)

println("Selected channels for onset_s: ", preview_onset.channel_selection)
fig_onset = plot_sort_variable_figure(preview_onset)
fig_onset
```
